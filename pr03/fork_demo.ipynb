```python
%%writefile fork_demo.py
import os

pid = os.fork()

if pid == 0:
    # Дочерний процесс
    print(
        f"Я ребёнок, PID={os.getpid()}, "
        f"PPID={os.getppid()}",
        flush=True,
    )
    os._exit(0)
else:
    # Родительский процесс
    child_pid, status = os.wait()
    exit_code = os.waitstatus_to_exitcode(status)

    print(f"Я родитель, PID={os.getpid()}")
    print(f"Дождался ребёнка PID={child_pid}")
    print(f"Код завершения ребёнка: {exit_code}")
```
